# Analisis de Datos de Joomla y PostgreSQL
Este cuaderno se conecta a la base de datos PostgreSQL de Joomla y realiza consultas sobre su estado y tablas.

```python
# 2. Configurar la conexion
import pandas as pd
import os

db_user = os.getenv('POSTGRES_USER', 'joomla_user')
db_pass = os.getenv('POSTGRES_PASSWORD', 'joomla_secure_password')
db_name = os.getenv('POSTGRES_DB', 'joomla')
db_host = 'database'

# Usamos la cadena de conexion directa (URI) requerida por las nuevas versiones de Pandas
uri = f'postgresql+psycopg2://{db_user}:{db_pass}@{db_host}:5432/{db_name}'
print("Conexion a PostgreSQL configurada correctamente.")
```

```python
# 3. Consultar tablas de Joomla y su tamano
query = """
SELECT relname as "Tabla", pg_size_pretty(pg_total_relation_size(relid)) as "Tamano"
FROM pg_catalog.pg_statio_user_tables
ORDER BY pg_total_relation_size(relid) DESC;
"""

df = pd.read_sql(query, uri)
display(df.head(10))
```
