Este cuaderno se conecta a la base de datos PostgreSQL de Joomla y realiza consultas sobre su estado y tablas.

```python
import psycopg2
import pandas as pd
import os
import warnings

# Ocultar la advertencia para conexiones directas
warnings.filterwarnings('ignore')

db_user = os.getenv('POSTGRES_USER', 'joomla_user')
db_pass = os.getenv('POSTGRES_PASSWORD', 'joomla_secure_password')
db_name = os.getenv('POSTGRES_DB', 'joomla')
db_host = 'database'

# Conexion pura y directa a PostgreSQL
conn = psycopg2.connect(
    host=db_host,
    database=db_name,
    user=db_user,
    password=db_pass,
    port=5432
)

query = """
SELECT relname as "Tabla", pg_size_pretty(pg_total_relation_size(relid)) as "Tamano"
FROM pg_catalog.pg_statio_user_tables
ORDER BY pg_total_relation_size(relid) DESC;
"""

# Leer los datos
df = pd.read_sql(query, conn)
display(df.head(10))

# Cerrar conexion por buenas practicas
conn.close()
```
