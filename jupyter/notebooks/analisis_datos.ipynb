Este cuaderno se conecta a la base de datos PostgreSQL de Joomla y realiza consultas sobre su estado y tablas.

```python
# 2. Conectar a PostgreSQL
import pandas as pd
import sqlalchemy
import os

db_user = os.getenv('POSTGRES_USER', 'joomla_user')
db_pass = os.getenv('POSTGRES_PASSWORD', 'joomla_secure_password')
db_name = os.getenv('POSTGRES_DB', 'joomla')
db_host = 'database'

engine = sqlalchemy.create_engine(f'postgresql://{db_user}:{db_pass}@{db_host}:5432/{db_name}')
print("Conexión a PostgreSQL establecida correctamente.")
```

```python
# 3. Consultar tablas y su tamaño
query = """
SELECT relname as "Tabla", pg_size_pretty(pg_total_relation_size(relid)) as "Tamaño"
FROM pg_catalog.pg_statio_user_tables
ORDER BY pg_total_relation_size(relid) DESC;
"""

df = pd.read_sql(query, engine)
display(df.head(10))
```
